## Recriar imagens e mascaras ausentes


```python
import numpy as np
from PIL import Image


def process_sprite_masked(
    raw_path,
    image_size=64,
    margin=4
):

    # Carregar sprite original
    rgba = Image.open(raw_path).convert("RGBA")

    # Separar RGB e alfa
    alpha = rgba.getchannel("A")

    # Encontrar região visível
    bbox = alpha.getbbox()

    if bbox is None:
        raise ValueError("Sprite sem pixels visíveis.")

    # Recortar imagem e máscara com a mesma região
    rgba = rgba.crop(bbox)
    alpha = alpha.crop(bbox)

    # Converter RGB para escala de cinza
    gray = rgba.convert("L")

    original_width, original_height = gray.size

    # Dimensão disponível
    available_size = image_size - 2 * margin

    # Preservar proporção
    scale = min(
        available_size / original_width,
        available_size / original_height
    )

    new_width = max(1, round(original_width * scale))
    new_height = max(1, round(original_height * scale))

    new_size = (new_width, new_height)

    # Redimensionar imagem e máscara
    gray = gray.resize(
        new_size,
        Image.Resampling.NEAREST
    )

    alpha = alpha.resize(
        new_size,
        Image.Resampling.NEAREST
    )

    # Criar canvases com fundo zero
    image_canvas = Image.new(
        "L",
        (image_size, image_size),
        0
    )

    mask_canvas = Image.new(
        "L",
        (image_size, image_size),
        0
    )

    # Calcular posição centralizada
    x_offset = (image_size - new_width) // 2
    y_offset = (image_size - new_height) // 2

    # Inserir imagem e máscara
    image_canvas.paste(
        gray,
        (x_offset, y_offset)
    )

    mask_canvas.paste(
        alpha,
        (x_offset, y_offset)
    )

    # Garantir que regiões transparentes não contenham intensidades
    image_array = np.array(image_canvas, dtype=np.uint8)
    mask_array = np.array(mask_canvas, dtype=np.uint8)

    image_array[mask_array == 0] = 0

    image_canvas = Image.fromarray(image_array)

    return image_canvas, mask_canvas
```

```python
# Recuperar somente arquivos intermediarios ausentes, sem alterar os metadados.
image_directory = Path(BASE_DIR) / "processed_masked"
mask_directory = Path(BASE_DIR) / "masks"
image_directory.mkdir(parents=True, exist_ok=True)
mask_directory.mkdir(parents=True, exist_ok=True)
source_metadata = pd.read_csv(Path(BASE_DIR) / "dataset_split.csv")
recreated = 0
for row in source_metadata.itertuples():
    filename = str(row.raw_path).replace("\\", "/").rsplit("/", 1)[-1]
    image_file = image_directory / filename
    mask_file = mask_directory / filename
    if image_file.is_file() and mask_file.is_file():
        continue
    raw_file = Path(BASE_DIR) / "raw" / filename
    if not raw_file.is_file():
        raise FileNotFoundError(f"Sprite original ausente: {raw_file}")
    processed_image, processed_mask = process_sprite_masked(raw_file, image_size=64, margin=4)
    if not image_file.is_file():
        processed_image.save(image_file)
    if not mask_file.is_file():
        processed_mask.save(mask_file)
    recreated += 1
print("Pares com arquivos recriados:", recreated)
```

```python
POKEMON_TYPES = [
    "normal",
    "fire",
    "water",
    "electric",
    "grass",
    "ice",
    "fighting",
    "poison",
    "ground",
    "flying",
    "psychic",
    "bug",
    "rock",
    "ghost",
    "dragon",
    "dark",
    "steel",
    "fairy"
]

STAT_COLUMNS = [
    "hp",
    "attack",
    "defense",
    "special_attack",
    "special_defense",
    "speed"
]

# Correspondência entre os nomes usados pela API
# e os nomes das nossas colunas
API_STAT_NAMES = {
    "hp": "hp",
    "attack": "attack",
    "defense": "defense",
    "special-attack": "special_attack",
    "special-defense": "special_defense",
    "speed": "speed"
}

print("Quantidade de tipos:", len(POKEMON_TYPES))
print("Quantidade de status:", len(STAT_COLUMNS))
```

```python
def fetch_pokemon_data(pokemon_id, session):

    url = f"https://pokeapi.co/api/v2/pokemon/{pokemon_id}/"

    response = None

    # Tentativas com tratamento de falhas temporárias
    for attempt in range(5):

        try:
            response = session.get(
                url,
                timeout=30
            )

            response.raise_for_status()
            break

        except requests.RequestException:

            if attempt == 4:
                raise

            time.sleep(2 ** attempt)

    data = response.json()

    # ID e nome retornados pela própria API
    api_id = int(data["id"])
    api_name = data["name"].lower()

    # Tipos ordenados pelo slot
    types = sorted(
        data["types"],
        key=lambda item: item["slot"]
    )

    types = [
        item["type"]["name"]
        for item in types
    ]

    # Status em um dicionário indexado pelo nome
    raw_stats = {
        item["stat"]["name"]: int(item["base_stat"])
        for item in data["stats"]
    }

    # Verificar se todos os seis status foram encontrados
    assert set(raw_stats.keys()) == set(API_STAT_NAMES.keys()), (
        f"Status inesperados para {api_name}: {raw_stats.keys()}"
    )

    record = {
        "id": api_id,
        "api_name": api_name,
        "types": "|".join(types)
    }

    for api_stat_name, column_name in API_STAT_NAMES.items():
        record[column_name] = raw_stats[api_stat_name]

    return record
```

```python
def normalize_name(name):
    return str(name).strip().lower().replace(" ", "-")


def validate_api_cache(cache_df, source_df):

    required = {
        "id",
        "api_name",
        "types",
        *STAT_COLUMNS
    }

    if not required.issubset(cache_df.columns):
        return False

    if cache_df["id"].duplicated().any():
        return False

    source_ids = set(source_df["id"].astype(int))
    cache_ids = set(cache_df["id"].astype(int))

    if source_ids != cache_ids:
        return False

    expected_names = (
        source_df
        .set_index("id")["pokemon_name"]
        .map(normalize_name)
    )

    cached_names = (
        cache_df
        .set_index("id")["api_name"]
        .map(normalize_name)
    )

    for pokemon_id in source_ids:

        if expected_names.loc[pokemon_id] != cached_names.loc[pokemon_id]:
            return False

    return True
```

```python
NUM_TYPES = len(POKEMON_TYPES)
NUM_STATS = len(STAT_COLUMNS)

TYPE_FEATURE_COLUMNS = [
    f"type_{pokemon_type}"
    for pokemon_type in POKEMON_TYPES
]

type_to_index = {
    pokemon_type: index
    for index, pokemon_type in enumerate(POKEMON_TYPES)
}


def encode_types(types_string):

    types = str(types_string).split("|")

    vector = np.zeros(
        NUM_TYPES,
        dtype=np.float32
    )

    for pokemon_type in types:

        pokemon_type = pokemon_type.strip()

        assert pokemon_type in type_to_index, (
            f"Tipo desconhecido: {pokemon_type}"
        )

        vector[type_to_index[pokemon_type]] = 1.0

    return vector
```

```python
class PokemonDataset(Dataset):

    def __init__(self, dataframe):

        self.df = dataframe.reset_index(
            drop=True
        ).copy()

        # Caminhos
        self.image_paths = self.df[
            "image_path"
        ].tolist()

        self.mask_paths = self.df[
            "mask_path"
        ].tolist()

        # Condições: matriz N x 24
        self.conditions = self.df[
            CONDITION_COLUMNS
        ].to_numpy(
            dtype=np.float32
        )

        assert self.conditions.shape == (
            len(self.df),
            CONDITION_DIM
        )

    def __len__(self):

        return len(self.df)

    def __getitem__(self, index):

        # Carregar imagem grayscale
        image = Image.open(
            self.image_paths[index]
        ).convert("L")

        image = np.asarray(
            image,
            dtype=np.float32
        ) / 255.0

        # Carregar máscara
        mask = Image.open(
            self.mask_paths[index]
        ).convert("L")

        mask = np.asarray(
            mask,
            dtype=np.uint8
        )

        # Converter para máscara binária 0/1
        mask = (mask > 127).astype(
            np.float32
        )

        assert image.shape == (64, 64)
        assert mask.shape == (64, 64)

        # Garantir background zerado
        image = image * mask

        # Tensores
        image = torch.from_numpy(
            image.copy()
        ).unsqueeze(0)

        mask = torch.from_numpy(
            mask.copy()
        ).unsqueeze(0)

        condition = torch.from_numpy(
            self.conditions[index].copy()
        )

        return image, mask, condition
```

```python
def evaluate_regions(model, loader):

    model.eval()

    foreground_error = 0.0
    background_error = 0.0

    foreground_pixels = 0.0
    background_pixels = 0.0

    with torch.no_grad():

        for images, masks, conditions in loader:

            images = images.to(DEVICE)
            masks = masks.to(DEVICE)
            conditions = conditions.to(DEVICE)

            reconstruction = model.reconstruct(
                images,
                conditions
            )

            error = torch.abs(
                images - reconstruction
            )

            background_mask = 1.0 - masks

            foreground_error += (
                error * masks
            ).sum().item()

            background_error += (
                error * background_mask
            ).sum().item()

            foreground_pixels += masks.sum().item()

            background_pixels += background_mask.sum().item()

    fg_mae = foreground_error / foreground_pixels
    bg_mae = background_error / background_pixels

    return {
        "foreground_mae": fg_mae,
        "background_mae": bg_mae
    }

def evaluate_silhouette_fixed(
    model,
    loader,
    threshold=0.20
):

    model.eval()

    all_iou = []
    all_dice = []

    with torch.no_grad():

        for images, masks, conditions in loader:

            images = images.to(DEVICE)
            masks = masks.to(DEVICE)
            conditions = conditions.to(DEVICE)

            reconstruction = model.reconstruct(
                images,
                conditions
            )

            predicted = reconstruction >= threshold
            target = masks >= 0.5

            intersection = (
                predicted & target
            ).sum(dim=(1, 2, 3)).float()

            union = (
                predicted | target
            ).sum(dim=(1, 2, 3)).float()

            pred_area = predicted.sum(
                dim=(1, 2, 3)
            ).float()

            target_area = target.sum(
                dim=(1, 2, 3)
            ).float()

            iou = intersection / union.clamp_min(1)

            dice = (
                2 * intersection
                / (pred_area + target_area).clamp_min(1)
            )

            all_iou.extend(iou.cpu().tolist())
            all_dice.extend(dice.cpu().tolist())

    return {
        "IoU": np.mean(all_iou),
        "Dice": np.mean(all_dice)
    }
```

# Arquitetura do Ultraball


```python
import torch.nn as nn
import torch.nn.functional as F
```

```python
def reparameterize(mu, logvar):

    std = torch.exp(0.5 * logvar)

    # DirectML nao possui Generator proprio: usar o RNG semeado da CPU.
    if std.device.type == "privateuseone":
        epsilon = torch.randn(std.shape, dtype=std.dtype, device="cpu").to(std.device)
    else:
        epsilon = torch.randn_like(std)

    z = mu + std * epsilon

    return z
```

```python
class SpatialEncoder16(nn.Module):

    def __init__(
        self,
        condition_dim=24,
        latent_channels=16
    ):

        super().__init__()

        self.features = nn.Sequential(

            # 64x64 -> 32x32
            nn.Conv2d(
                1 + condition_dim,
                64,
                kernel_size=4,
                stride=2,
                padding=1
            ),
            nn.BatchNorm2d(64),
            nn.LeakyReLU(0.2, inplace=True),

            # 32x32 -> 16x16
            nn.Conv2d(
                64,
                128,
                kernel_size=4,
                stride=2,
                padding=1
            ),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),

            # Refinamento espacial 16x16
            nn.Conv2d(
                128,
                192,
                kernel_size=3,
                stride=1,
                padding=1
            ),
            nn.BatchNorm2d(192),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(
                192,
                192,
                kernel_size=3,
                stride=1,
                padding=1
            ),
            nn.BatchNorm2d(192),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(
                192,
                192,
                kernel_size=3,
                stride=1,
                padding=1
            ),
            nn.BatchNorm2d(192),
            nn.LeakyReLU(0.2, inplace=True),

            # Preparar representação latente
            nn.Conv2d(
                192,
                128,
                kernel_size=1
            ),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True)
        )

        # Projeções espaciais para os parâmetros da distribuição
        self.mu_head = nn.Conv2d(
            128,
            latent_channels,
            kernel_size=1
        )

        self.logvar_head = nn.Conv2d(
            128,
            latent_channels,
            kernel_size=1
        )

        # Inicialização pequena da média
        nn.init.xavier_uniform_(
            self.mu_head.weight,
            gain=0.1
        )

        nn.init.zeros_(
            self.mu_head.bias
        )

        # Inicializar logvar em zero:
        # distribuição inicial próxima de N(0, I)
        nn.init.zeros_(
            self.logvar_head.weight
        )

        nn.init.zeros_(
            self.logvar_head.bias
        )

    def forward(self, x, c):

        batch_size = x.size(0)

        height = x.size(2)
        width = x.size(3)

        # Expandir condições espacialmente
        c_map = c.unsqueeze(-1).unsqueeze(-1)

        c_map = c_map.expand(
            batch_size,
            c.size(1),
            height,
            width
        )

        # Concatenar imagem e condições
        x = torch.cat(
            [x, c_map],
            dim=1
        )

        # Extração espacial
        features = self.features(x)

        # Parâmetros da distribuição latente
        mu = self.mu_head(features)

        raw_logvar = self.logvar_head(features)

        # Estabilização suave
        LOGVAR_LIMIT = 6.0

        logvar = LOGVAR_LIMIT * torch.tanh(
            raw_logvar / LOGVAR_LIMIT
        )

        return mu, logvar
```

```python
class SpatialDecoder16(nn.Module):

    def __init__(
        self,
        latent_channels=16,
        condition_dim=24
    ):

        super().__init__()

        input_channels = latent_channels + condition_dim

        self.initial = nn.Sequential(

            nn.Conv2d(
                input_channels,
                192,
                kernel_size=3,
                stride=1,
                padding=1
            ),
            nn.BatchNorm2d(192),
            nn.ReLU(inplace=True)
        )

        # Refinamento em 16x16
        self.refine = nn.Sequential(

            nn.Conv2d(
                192,
                192,
                kernel_size=3,
                padding=1
            ),
            nn.BatchNorm2d(192),
            nn.ReLU(inplace=True),

            nn.Conv2d(
                192,
                192,
                kernel_size=3,
                padding=1
            ),
            nn.BatchNorm2d(192),
            nn.ReLU(inplace=True),

            nn.Conv2d(
                192,
                128,
                kernel_size=3,
                padding=1
            ),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )

        # 16x16 -> 32x32
        self.up32 = nn.Sequential(

            nn.Upsample(
                scale_factor=2,
                mode="nearest"
            ),

            nn.Conv2d(
                128,
                96,
                kernel_size=3,
                padding=1
            ),
            nn.BatchNorm2d(96),
            nn.ReLU(inplace=True),

            nn.Conv2d(
                96,
                96,
                kernel_size=3,
                padding=1
            ),
            nn.BatchNorm2d(96),
            nn.ReLU(inplace=True)
        )

        # 32x32 -> 64x64
        self.up64 = nn.Sequential(

            nn.Upsample(
                scale_factor=2,
                mode="nearest"
            ),

            nn.Conv2d(
                96,
                64,
                kernel_size=3,
                padding=1
            ),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),

            nn.Conv2d(
                64,
                32,
                kernel_size=3,
                padding=1
            ),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )

        # Refinamento final e saída grayscale
        self.output = nn.Sequential(

            nn.Conv2d(
                32,
                1,
                kernel_size=3,
                padding=1
            ),

            nn.Sigmoid()
        )

    def forward(self, z, c):

        batch_size = z.size(0)

        height = z.size(2)
        width = z.size(3)

        # Expandir condições para 16x16
        c_map = c.unsqueeze(-1).unsqueeze(-1)

        c_map = c_map.expand(
            batch_size,
            c.size(1),
            height,
            width
        )

        # Concatenar latente e condições
        x = torch.cat(
            [z, c_map],
            dim=1
        )

        x = self.initial(x)

        x = self.refine(x)

        x = self.up32(x)

        x = self.up64(x)

        reconstruction = self.output(x)

        return reconstruction
```

```python
class CVAESpatial16(nn.Module):

    def __init__(
        self,
        latent_channels=16,
        condition_dim=24
    ):

        super().__init__()

        self.latent_channels = latent_channels
        self.condition_dim = condition_dim

        self.encoder = SpatialEncoder16(
            condition_dim=condition_dim,
            latent_channels=latent_channels
        )

        self.decoder = SpatialDecoder16(
            latent_channels=latent_channels,
            condition_dim=condition_dim
        )

    def forward(self, x, c):

        mu, logvar = self.encoder(x, c)

        z = reparameterize(
            mu,
            logvar
        )

        reconstruction = self.decoder(
            z,
            c
        )

        return reconstruction, mu, logvar

    def reconstruct(self, x, c):

        # Reconstrução determinística
        mu, logvar = self.encoder(x, c)

        reconstruction = self.decoder(
            mu,
            c
        )

        return reconstruction

    def generate(self, c, z=None):

        batch_size = c.size(0)

        if z is None:

            z = torch.randn(
                batch_size,
                self.latent_channels,
                16,
                16,
                device="cpu" if c.device.type == "privateuseone" else c.device,
                dtype=c.dtype
            ).to(c.device)

        return self.decoder(
            z,
            c
        )
```

```python
# Experimento atual: 4 canais. Use 16 para repetir a referencia SSIM anterior.
LATENT_CHANNELS = 4
LATENT_SIZE = 16

CONDITION_DIM = 24

LATENT_ELEMENTS = (
    LATENT_CHANNELS
    * LATENT_SIZE
    * LATENT_SIZE
)

REFERENCE_LATENT_DIM = 128

# Fixar o divisor da referencia evita quadruplicar o peso da KL ao reduzir canais.
KL_NORMALIZATION = (
    (16 * LATENT_SIZE * LATENT_SIZE)
    / REFERENCE_LATENT_DIM
)

print("Canais latentes:", LATENT_CHANNELS)
print("Resolução latente:", LATENT_SIZE, "x", LATENT_SIZE)
print("Total de variáveis latentes:", LATENT_ELEMENTS)
print("Fator de normalização da KL:", KL_NORMALIZATION)
```

```python
def spatial_kl_divergence(mu, logvar):

    kl_per_sample = -0.5 * (
        1
        + logvar
        - mu.pow(2)
        - logvar.exp()
    ).sum(dim=(1, 2, 3))

    kl_raw = kl_per_sample.mean()

    kl_normalized = (
        kl_raw / KL_NORMALIZATION
    )

    return kl_normalized, kl_raw
```

```python
def ssim_error_map(x, reconstruction, window_size=7, sigma=1.5):
    # SSIM local implementada aqui, sem rede ou biblioteca pre-treinada.
    # Intensidades [0,1]; janela gaussiana 7x7; padding refletido.
    if x.shape != reconstruction.shape or x.ndim != 4:
        raise ValueError("SSIM espera tensores NCHW com formas iguais.")
    if window_size < 3 or window_size % 2 == 0 or sigma <= 0:
        raise ValueError("Janela deve ser impar >=3 e sigma positivo.")
    radius = window_size // 2
    if min(x.shape[-2:]) <= radius:
        raise ValueError("Imagem pequena demais para a janela SSIM.")
    coordinates = torch.arange(window_size, device=x.device, dtype=x.dtype) - radius
    gaussian = torch.exp(-(coordinates ** 2) / (2.0 * sigma ** 2))
    gaussian = gaussian / gaussian.sum()
    kernel = (gaussian[:, None] * gaussian[None, :])[None, None]
    channels = x.shape[1]
    kernel = kernel.expand(channels, 1, window_size, window_size).contiguous()
    def local_mean(value):
        return F.conv2d(F.pad(value, (radius, radius, radius, radius), mode="reflect"),
                        kernel, groups=channels)
    # Mesmo filtro para medias, segundos momentos e produto cruzado.
    moments = local_mean(torch.cat([x, reconstruction, x*x,
                                    reconstruction*reconstruction, x*reconstruction], dim=0))
    mean_x, mean_y, second_x, second_y, cross = moments.chunk(5, dim=0)
    variance_x = (second_x - mean_x.square()).clamp_min(0.0)
    variance_y = (second_y - mean_y.square()).clamp_min(0.0)
    covariance = cross - mean_x * mean_y
    c1, c2 = 0.01 ** 2, 0.03 ** 2
    numerator = (2 * mean_x * mean_y + c1) * (2 * covariance + c2)
    denominator = (mean_x.square() + mean_y.square() + c1) * (variance_x + variance_y + c2)
    similarity = numerator / denominator
    # Mapa de erro [0,1]; a mascara sera aplicada depois, no centro de cada janela.
    return (1.0 - similarity.clamp(-1.0, 1.0)) / 2.0


LAMBDA_BG = 0.1


def cvae_spatial_loss(
    x,
    reconstruction,
    mask,
    mu,
    logvar,
    beta=1.0,
    lambda_bg=0.1,
    reconstruction_kind="l1"
):

    if reconstruction_kind == "l1":
        pixel_error = torch.abs(x - reconstruction)
    elif reconstruction_kind == "bce":
        # Alvos em tons de cinza [0,1], sem binarizar as imagens.
        # O decoder ja aplica sigmoid. Clamp evita log(0).
        prediction = reconstruction.clamp(1e-6, 1.0 - 1e-6)
        pixel_error = F.binary_cross_entropy(prediction, x, reduction="none")
    elif reconstruction_kind == "ssim":
        pixel_error = ssim_error_map(x, reconstruction)
    else:
        raise ValueError("Perda de reconstrucao desconhecida: " + reconstruction_kind)

    # Foreground
    foreground_loss = (
        pixel_error * mask
    ).sum(dim=(1, 2, 3)).mean()

    # Background
    background_loss = (
        pixel_error * (1.0 - mask)
    ).sum(dim=(1, 2, 3)).mean()

    # KL espacial normalizada
    kl_loss, kl_raw = spatial_kl_divergence(
        mu,
        logvar
    )

    # Reconstrução ponderada
    reconstruction_loss = (
        foreground_loss
        + lambda_bg * background_loss
    )

    # Loss total
    total_loss = (
        reconstruction_loss
        + beta * kl_loss
    )

    return (
        total_loss,
        foreground_loss,
        background_loss,
        kl_loss,
        kl_raw
    )

def masked_sobel_loss(x, reconstruction, mask):
    # Filtros fixos: nenhum peso pre-treinado ou parametro adicional.
    sobel_x = x.new_tensor([[-1., 0., 1.], [-2., 0., 2.], [-1., 0., 1.]]).view(1, 1, 3, 3) / 8.0
    sobel_y = sobel_x.transpose(2, 3).contiguous()
    target_x = F.conv2d(x, sobel_x, padding=1)
    target_y = F.conv2d(x, sobel_y, padding=1)
    predicted_x = F.conv2d(reconstruction, sobel_x, padding=1)
    predicted_y = F.conv2d(reconstruction, sobel_y, padding=1)
    # Dilatar 1 pixel inclui a transicao entre Pokemon e fundo no contorno.
    edge_region = F.max_pool2d(mask, kernel_size=3, stride=1, padding=1)
    error = 0.5 * ((predicted_x - target_x).abs() + (predicted_y - target_y).abs())
    return (error * edge_region).sum(dim=(1, 2, 3)).mean()
```

### Treinamento: SSIM + mascara alfa + Sobel

`SPATIAL_EXPERIMENT = "ssim_bordas_beta2"` seleciona:

`erro SSIM no Pokemon + 0,1 * erro SSIM no fundo + 0,1 * Sobel + beta * KL normalizada`

SSIM substitui a L1 no objetivo de treino. A implementacao explicita calcula medias, variancias e covariancia locais com uma janela gaussiana 7x7 (sigma 1,5), padding refletido e constantes C1=0,01^2 e C2=0,03^2 para tons de cinza [0,1]. Usamos o mapa `(1 - SSIM) / 2`, somado por imagem e promediado no lote. A mascara alfa pondera o centro de cada janela; as vizinhancas do contorno incluem pixels do fundo. Este e um objetivo local adaptado a sprites, nao a implementacao oficial completa. Base: [Wang et al., 2004](https://www.cns.nyu.edu/~lcv/ssim/).

Nao ha redes pre-treinadas nem novas dependencias. Exceto pela reducao para 4 canais latentes (configuravel na celula do latente), a arquitetura e os dados, semente 42, warm-up de 40 e limite de 150 epocas permanecem iguais. Sobel continua com peso 0,1. SSIM nao garante reduzir borrado; sua escala muda o peso efetivo relativo de Sobel e KL.

Para comparar, `original_bordas` preserva L1 + Sobel e `bce_bordas` preserva BCE + Sobel. Os arquivos novos recebem sufixo `_ssim`, e o ZIP registra os parametros da SSIM. Execute todas as celulas do zero.

Scheduler, parada antecipada e melhor checkpoint continuam usando **L1 mascarada + KL com beta final fixo**, sem Sobel, como criterio comum. L1 e somente metrica no treino SSIM, sem gradiente. `val_selection_score` e `best_val_loss` registram esse criterio; `val_total` registra o objetivo SSIM + Sobel + KL. Os campos foreground/background da perda selecionada passam a medir erro SSIM. Compare `val_l1_foreground`, `val_l1_background`, KL, Sobel e galerias com mesmas condicoes e latentes; nao compare diretamente os totais SSIM, L1 e BCE.

Neste teste de 4 canais, mantemos o divisor da KL em 32, igual ao modelo anterior de 16 canais; nao usamos 1024/128=8, pois isso quadruplicaria o coeficiente da KL bruta. As metricas de KL bruta e normalizada continuam registradas. A configuracao de canais deve ser alterada antes de criar o modelo, executando o notebook do inicio.

O teste atual `ssim_bordas_beta2` usa **beta final 2**, comparado ao beta final 1 de `ssim_bordas`. Mantemos 4 canais 16x16, SSIM, mascara, Sobel 0,1, divisor KL 32, dados, semente, batch, otimizador, scheduler e regras de parada. Na epoca 1 beta=0,05, na 20 beta=1 e na 40 beta=2. Os nomes incluem `_kl_beta2_warmup40_`, preservando os resultados beta1.

O criterio de selecao continua L1 mascarada + KL com o beta final da configuracao. Portanto **nao compare diretamente `val_selection_score` entre beta1 e beta2**. Compare L1, Sobel, KL bruta, geracao com condicoes e sementes fixas e a metrica `val_l1_plus_kl_beta1` (registrada com beta1 fixo apenas para comparacao; nao controla treino ou selecao). Este teste pode perder detalhes de reconstrucao sem melhorar a geracao; isso deve ser avaliado pelos resultados.


```python
def train_one_epoch_spatial(
    model,
    loader,
    optimizer,
    beta,
    lambda_bg,
    beta_reference=1.0,
    lambda_edge=0.0,
    reconstruction_kind="l1"
):

    model.train()

    total_fg = 0.0
    total_bg = 0.0

    total_kl = 0.0
    total_kl_raw = 0.0

    total_edge = 0.0
    total_objective = 0.0

    max_abs_logvar = 0.0
    num_samples = 0

    for images, masks, conditions in loader:

        images = images.to(DEVICE)
        masks = masks.to(DEVICE)
        conditions = conditions.to(DEVICE)

        optimizer.zero_grad(set_to_none=True)

        reconstruction, mu, logvar = model(
            images,
            conditions
        )

        if not torch.isfinite(logvar).all():
            raise FloatingPointError(
                "Logvar contém valores não finitos."
            )

        max_abs_logvar = max(
            max_abs_logvar,
            logvar.abs().max().item()
        )

        (
            loss,
            fg_loss,
            bg_loss,
            kl_loss,
            kl_raw
        ) = cvae_spatial_loss(
            images,
            reconstruction,
            masks,
            mu,
            logvar,
            beta=beta,
            lambda_bg=lambda_bg,
            reconstruction_kind=reconstruction_kind
        )

        edge_loss = masked_sobel_loss(images, reconstruction, masks)
        loss = loss + lambda_edge * edge_loss

        if not torch.isfinite(loss):
            raise FloatingPointError(
                "Loss não finita durante o treinamento."
            )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=GRAD_CLIP
        )

        optimizer.step()

        batch_size = images.size(0)

        total_fg += fg_loss.item() * batch_size
        total_bg += bg_loss.item() * batch_size

        total_kl += kl_loss.item() * batch_size
        total_kl_raw += kl_raw.item() * batch_size

        total_edge += edge_loss.item() * batch_size
        total_objective += loss.item() * batch_size

        num_samples += batch_size

    avg_fg = total_fg / num_samples
    avg_bg = total_bg / num_samples

    avg_kl = total_kl / num_samples
    avg_kl_raw = total_kl_raw / num_samples

    avg_edge = total_edge / num_samples
    avg_objective = total_objective / num_samples

    # Objetivo com o beta final fixo como referencia.
    avg_total_no_edge = (
        avg_fg
        + lambda_bg * avg_bg
        + beta_reference * avg_kl
    )

    avg_total = avg_total_no_edge + lambda_edge * avg_edge

    return {
        "total": avg_total,
        "total_no_edge": avg_total_no_edge,
        "edge": avg_edge,
        "edge_weighted": lambda_edge * avg_edge,
        "objective": avg_objective,

        "foreground": avg_fg,
        "background": avg_bg,

        "kl": avg_kl,
        "kl_raw": avg_kl_raw,
        "kl_weighted": beta * avg_kl,
        "kl_reference": beta_reference * avg_kl,

        "max_abs_logvar": max_abs_logvar
    }
```

```python
def evaluate_spatial(
    model,
    loader,
    lambda_bg=0.1,
    beta=1.0,
    lambda_edge=0.0,
    reconstruction_kind="l1"
):

    model.eval()

    total_fg = 0.0
    total_bg = 0.0

    total_kl = 0.0
    total_kl_raw = 0.0
    total_edge = 0.0
    total_l1_fg = 0.0
    total_l1_bg = 0.0

    num_samples = 0

    with torch.no_grad():

        for images, masks, conditions in loader:

            images = images.to(DEVICE)
            masks = masks.to(DEVICE)
            conditions = conditions.to(DEVICE)

            # Encoder
            mu, logvar = model.encoder(
                images,
                conditions
            )

            # Decoder determinístico
            reconstruction = model.decoder(
                mu,
                conditions
            )

            (
                loss,
                fg_loss,
                bg_loss,
                kl_loss,
                kl_raw
            ) = cvae_spatial_loss(
                images,
                reconstruction,
                masks,
                mu,
                logvar,
                beta=beta,
                lambda_bg=lambda_bg,
                reconstruction_kind=reconstruction_kind
            )

            l1_error = (images - reconstruction).abs()
            total_l1_fg += (l1_error * masks).sum(dim=(1, 2, 3)).mean().item() * images.size(0)
            total_l1_bg += (l1_error * (1.0 - masks)).sum(dim=(1, 2, 3)).mean().item() * images.size(0)
            edge_loss = masked_sobel_loss(images, reconstruction, masks)
            total_edge += edge_loss.item() * images.size(0)

            batch_size = images.size(0)

            total_fg += fg_loss.item() * batch_size
            total_bg += bg_loss.item() * batch_size

            total_kl += kl_loss.item() * batch_size
            total_kl_raw += kl_raw.item() * batch_size

            num_samples += batch_size

    avg_fg = total_fg / num_samples
    avg_bg = total_bg / num_samples

    avg_kl = total_kl / num_samples
    avg_kl_raw = total_kl_raw / num_samples

    avg_edge = total_edge / num_samples
    avg_total_no_edge = (
        avg_fg
        + lambda_bg * avg_bg
        + beta * avg_kl
    )

    avg_total = avg_total_no_edge + lambda_edge * avg_edge

    return {
        "l1_foreground": total_l1_fg / num_samples,
        "l1_background": total_l1_bg / num_samples,
        "selection_score": (total_l1_fg + lambda_bg * total_l1_bg) / num_samples + beta * avg_kl,
        "total": avg_total,
        "total_no_edge": avg_total_no_edge,
        "edge": avg_edge,
        "edge_weighted": lambda_edge * avg_edge,

        "foreground": avg_fg,
        "background": avg_bg,

        "kl": avg_kl,
        "kl_raw": avg_kl_raw,
        "kl_weighted": beta * avg_kl
    }
```

```python
# Semente
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

# Reinicializar o modelo
model_spatial = CVAESpatial16(
    latent_channels=LATENT_CHANNELS,
    condition_dim=CONDITION_DIM
).to(DEVICE)

# DataLoader reproduzível
generator = torch.Generator()
generator.manual_seed(SEED)

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    generator=generator
)

# Hiperparâmetros
EPOCHS = 150

LEARNING_RATE = 1e-3

# Use "ssim_bordas" para repetir SSIM com beta final 1.
# Este teste muda somente beta final para 2 (warm-up continua 40).
SPATIAL_EXPERIMENT = "ssim_bordas_beta2"
SPATIAL_KL_CONFIGS = {
    "original": {"beta_max": 1.0, "warmup_epochs": 40},
    "kl_reforcada": {"beta_max": 2.0, "warmup_epochs": 80},
    "kl_beta2_warmup40": {"beta_max": 2.0, "warmup_epochs": 40},
    "original_bordas": {"beta_max": 1.0, "warmup_epochs": 40, "lambda_edge": 0.1},
    "bce_bordas": {"beta_max": 1.0, "warmup_epochs": 40, "lambda_edge": 0.1, "reconstruction_kind": "bce"},
    "ssim_bordas": {"beta_max": 1.0, "warmup_epochs": 40, "lambda_edge": 0.1, "reconstruction_kind": "ssim"},
    "ssim_bordas_beta2": {"beta_max": 2.0, "warmup_epochs": 40, "lambda_edge": 0.1, "reconstruction_kind": "ssim"}
}
kl_config = SPATIAL_KL_CONFIGS[SPATIAL_EXPERIMENT]
BETA_MAX = kl_config["beta_max"]
BETA_WARMUP_EPOCHS = kl_config["warmup_epochs"]
LAMBDA_EDGE = kl_config.get("lambda_edge", 0.0)
RECONSTRUCTION_KIND = kl_config.get("reconstruction_kind", "l1")
RECONSTRUCTION_DETAILS = ({"window_size": 7, "sigma": 1.5, "data_range": 1.0,
                           "k1": 0.01, "k2": 0.03, "padding": "reflect",
                           "error_map": "(1-ssim)/2", "reduction": "masked_sum_per_image_mean_batch"}
                          if RECONSTRUCTION_KIND == "ssim" else {})

def spatial_beta(epoch, beta_max, warmup_epochs):
    # epoch comeca em zero. O teto e beta_max, mesmo quando maior que 1.
    if epoch < 0 or beta_max < 0 or warmup_epochs <= 0:
        raise ValueError("Epoca e beta devem ser nao negativos; warm-up deve ser positivo.")
    return beta_max * min(1.0, (epoch + 1) / warmup_epochs)

SPATIAL_RUN_NAME = f"cvae_spatial_{LATENT_CHANNELS}x{LATENT_SIZE}x{LATENT_SIZE}"
if SPATIAL_EXPERIMENT != "original":
    SPATIAL_RUN_NAME += f"_kl_beta{BETA_MAX:g}_warmup{BETA_WARMUP_EPOCHS}"
if LAMBDA_EDGE > 0:
    SPATIAL_RUN_NAME += f"_sobel{LAMBDA_EDGE:g}".replace(".", "p")

if RECONSTRUCTION_KIND != "l1":
    SPATIAL_RUN_NAME += "_" + RECONSTRUCTION_KIND

LAMBDA_BG = 0.1

PATIENCE = 20
SCHEDULER_PATIENCE = 7

MIN_DELTA = 1e-4
MIN_LR = 1e-6
GRAD_CLIP = 5.0

optimizer_spatial = torch.optim.Adam(
    model_spatial.parameters(),
    lr=LEARNING_RATE
)

scheduler_spatial = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_spatial,
    mode="min",
    factor=0.5,
    patience=SCHEDULER_PATIENCE,
    min_lr=MIN_LR
)

print("Treinamento espacial configurado!")
print("Latent shape:", (LATENT_CHANNELS, LATENT_SIZE, LATENT_SIZE))
print("Variáveis latentes:", LATENT_ELEMENTS)
print("KL normalization:", KL_NORMALIZATION)
print("Lambda background:", LAMBDA_BG)

print("Experimento:", SPATIAL_EXPERIMENT)
print("Beta final:", BETA_MAX, "| Warm-up:", BETA_WARMUP_EPOCHS)
print("Coeficiente final da KL bruta:", BETA_MAX / KL_NORMALIZATION)
print("Beta nas epocas 1, 40, 80:", [spatial_beta(e - 1, BETA_MAX, BETA_WARMUP_EPOCHS) for e in [1, 40, 80]])

print("Dispositivo dos pesos do modelo:", next(model_spatial.parameters()).device)

print("Reconstrucao:", RECONSTRUCTION_KIND)
print("Lambda bordas:", LAMBDA_EDGE)
print("Selecao do checkpoint: L1 mascarada + KL na validacao (objetivo comum).")
```

```python
def checkpoint_to_cpu(value):
    if torch.is_tensor(value):
        return value.detach().cpu()
    if isinstance(value, dict):
        return {key: checkpoint_to_cpu(item) for key, item in value.items()}
    if isinstance(value, list):
        return [checkpoint_to_cpu(item) for item in value]
    if isinstance(value, tuple):
        return tuple(checkpoint_to_cpu(item) for item in value)
    return value

CHECKPOINT_SPATIAL_PATH = os.path.join(
    BASE_DIR,
    f"{SPATIAL_RUN_NAME}_best.pt"
)

HISTORY_SPATIAL_PATH = os.path.join(
    BASE_DIR,
    f"{SPATIAL_RUN_NAME}_history.csv"
)

history_spatial = []

best_val_loss = float("inf")
best_epoch = 0

epochs_without_improvement = 0

for epoch in range(EPOCHS):

    # Beta warm-up linear
    beta = spatial_beta(epoch, BETA_MAX, BETA_WARMUP_EPOCHS)

    # Treinamento
    train_metrics = train_one_epoch_spatial(
        model_spatial,
        train_loader,
        optimizer_spatial,
        beta,
        LAMBDA_BG,
        beta_reference=BETA_MAX,
        lambda_edge=LAMBDA_EDGE,
        reconstruction_kind=RECONSTRUCTION_KIND
    )

    # Objetivo com o beta final fixo como referencia.
    val_metrics = evaluate_spatial(
        model_spatial,
        val_loader,
        lambda_bg=LAMBDA_BG,
        beta=BETA_MAX,
        lambda_edge=LAMBDA_EDGE,
        reconstruction_kind=RECONSTRUCTION_KIND
    )

    # Scheduler
    # Mesmo criterio da referencia: L1 com mascara + KL, sem o termo adicional.
    selection_score = val_metrics["selection_score"]
    scheduler_spatial.step(selection_score)

    current_lr = optimizer_spatial.param_groups[0]["lr"]

    # Histórico
    epoch_data = {
        "epoch": epoch + 1,
        "beta": beta,
        "lr": current_lr,

        "train_total": train_metrics["total"],
        "train_total_no_edge": train_metrics["total_no_edge"],
        "train_edge": train_metrics["edge"],
        "train_edge_weighted": train_metrics["edge_weighted"],
        "train_objective": train_metrics["objective"],

        "train_foreground": train_metrics["foreground"],
        "train_background": train_metrics["background"],

        "train_kl": train_metrics["kl"],
        "train_kl_raw": train_metrics["kl_raw"],
        "train_kl_weighted": train_metrics["kl_weighted"],
        "train_kl_reference": train_metrics["kl_reference"],

        "train_max_abs_logvar": train_metrics["max_abs_logvar"],

        "val_total": val_metrics["total"],
        "val_selection_score": selection_score,
        "val_l1_plus_kl_beta1": val_metrics["l1_foreground"] + LAMBDA_BG * val_metrics["l1_background"] + val_metrics["kl"],
        "val_l1_foreground": val_metrics["l1_foreground"],
        "val_l1_background": val_metrics["l1_background"],
        "val_total_no_edge": val_metrics["total_no_edge"],
        "val_edge": val_metrics["edge"],
        "val_edge_weighted": val_metrics["edge_weighted"],

        "val_foreground": val_metrics["foreground"],
        "val_background": val_metrics["background"],

        "val_kl": val_metrics["kl"],
        "val_kl_raw": val_metrics["kl_raw"],
        "val_kl_weighted": val_metrics["kl_weighted"]
    }

    history_spatial.append(epoch_data)

    # Salvar melhor checkpoint
    if selection_score < best_val_loss - MIN_DELTA:

        best_val_loss = float(selection_score)
        best_epoch = epoch + 1

        epochs_without_improvement = 0

        torch.save(
            {
                "epoch": best_epoch,

                "model_state_dict": {key: value.detach().cpu() for key, value in model_spatial.state_dict().items()},
                "optimizer_state_dict": checkpoint_to_cpu(optimizer_spatial.state_dict()),
                "device_name": DEVICE_NAME,
                "torch_version": str(torch.__version__),

                "best_val_loss": float(best_val_loss),
                "best_val_l1_plus_kl_beta1": float(val_metrics["l1_foreground"] + LAMBDA_BG * val_metrics["l1_background"] + val_metrics["kl"]),
                "best_val_total_with_edge": float(val_metrics["total"]),
                "selection_metric": "val_l1_masked_plus_kl",
                "reconstruction_kind": RECONSTRUCTION_KIND,
                "reconstruction_details": RECONSTRUCTION_DETAILS,
                "lambda_edge": float(LAMBDA_EDGE),
                "edge_loss": "sobel_l1_dilated_alpha_mask",

                "latent_channels": LATENT_CHANNELS,
                "latent_size": LATENT_SIZE,
                "latent_elements": LATENT_ELEMENTS,

                "condition_dim": CONDITION_DIM,

                "kl_normalization": float(KL_NORMALIZATION),
                "kl_normalization_policy": "fixed_reference_16_channels",
                "reference_latent_dim": REFERENCE_LATENT_DIM,

                "lambda_bg": float(LAMBDA_BG),

                "experiment": SPATIAL_EXPERIMENT,
                "validation_beta": float(BETA_MAX),
                "beta_at_epoch": float(beta),
                "beta_max": float(BETA_MAX),
                "beta_warmup_epochs": BETA_WARMUP_EPOCHS,

                "logvar_limit": 6.0,

                "seed": SEED
            },
            CHECKPOINT_SPATIAL_PATH
        )

        status = "Melhor modelo salvo!"

    else:

        status = ""

        if epoch + 1 >= BETA_WARMUP_EPOCHS:
            epochs_without_improvement += 1

    # Salvar histórico
    pd.DataFrame(history_spatial).to_csv(
        HISTORY_SPATIAL_PATH,
        index=False
    )

    print(
        f"Epoch [{epoch + 1:03d}/{EPOCHS}] "
        f"Beta: {beta:.3f} | "
        f"LR: {current_lr:.6f} | "

        f"Train FG: {train_metrics['foreground']:.2f} | "
        f"Train BG: {train_metrics['background']:.2f} | "

        f"KL norm: {train_metrics['kl']:.3f} | "
        f"Beta*KL: {train_metrics['kl_weighted']:.3f} | "
        f"KL raw: {train_metrics['kl_raw']:.2f} | "

        f"Val {RECONSTRUCTION_KIND.upper()} FG: {val_metrics['foreground']:.2f} | "
        f"Val BG: {val_metrics['background']:.2f} | "

        f"Val KL: {val_metrics['kl']:.3f} | "
        f"Val edge: {val_metrics['edge']:.2f} | "
        f"Val L1+KL: {selection_score:.3f} | "
        f"{status}"
    )

    # Early stopping
    if (
        epoch + 1 >= BETA_WARMUP_EPOCHS
        and epochs_without_improvement >= PATIENCE
    ):

        print(
            f"\nEarly stopping na época {epoch + 1}."
        )

        break


print("\nTreinamento finalizado!")
print("Melhor época:", best_epoch)
print("Melhor validação:", best_val_loss)
print("Checkpoint:", CHECKPOINT_SPATIAL_PATH)
```

```python
def inspect_spatial_latent(model, loader):

    model.eval()

    all_mu = []
    all_logvar = []

    with torch.no_grad():

        for images, masks, conditions in loader:

            images = images.to(DEVICE)
            conditions = conditions.to(DEVICE)

            mu, logvar = model.encoder(
                images,
                conditions
            )

            all_mu.append(mu.cpu())
            all_logvar.append(logvar.cpu())

    mu = torch.cat(all_mu, dim=0)
    logvar = torch.cat(all_logvar, dim=0)

    std = torch.exp(0.5 * logvar)

    # KL por posição espacial e canal
    kl_map = -0.5 * (
        1 + logvar
        - mu.pow(2)
        - logvar.exp()
    )

    # Média por canal, batch e posições espaciais
    kl_per_channel = kl_map.mean(
        dim=(0, 2, 3)
    )

    print("Formato do mu:", mu.shape)

    print("\nDistribuição latente:")
    print("Mu mean:", mu.mean().item())
    print("Mu std:", mu.std().item())

    print("\nDesvio padrão:")
    print("Std mean:", std.mean().item())
    print("Std min:", std.min().item())
    print("Std max:", std.max().item())

    print("\nKL média por canal:")

    for i, value in enumerate(kl_per_channel):

        print(
            f"Canal {i + 1:02d}: {value.item():.6f}"
        )

    return mu, logvar, kl_per_channel
```

```python
# Medir atividade e efeito de zerar cada canal no melhor checkpoint.
# Usa apenas validacao; nao treina e nao modifica os pesos.
def analyze_latent_channels(model, loader):
    model.eval()
    mus, logvars = [], []
    with torch.no_grad():
        for images, masks, conditions in loader:
            mu, logvar = model.encoder(images.to(DEVICE), conditions.to(DEVICE))
            mus.append(mu.cpu())
            logvars.append(logvar.cpu())
    means = torch.cat(mus)
    logvars = torch.cat(logvars)
    kl_by_channel = (0.5 * (means.square() + logvars.exp() - 1 - logvars)).sum((2, 3)).mean(0)
    total_kl = kl_by_channel.sum().item()
    variances = means.var(0, unbiased=False).mean((1, 2))
    channels = model.latent_channels
    impacts = np.zeros(channels)
    foreground_deltas = np.zeros(channels)
    samples = 0
    with torch.no_grad():
        for images, masks, conditions in loader:
            images, masks, conditions = images.to(DEVICE), masks.to(DEVICE), conditions.to(DEVICE)
            mean, _ = model.encoder(images, conditions)
            baseline = model.decoder(mean, conditions)
            baseline_error = ((baseline - images).abs() * masks).sum((1, 2, 3))
            for channel in range(channels):
                altered = mean.clone()
                altered[:, channel] = 0
                changed = model.decoder(altered, conditions)
                impacts[channel] += (changed - baseline).abs().mean((1, 2, 3)).sum().item()
                foreground_deltas[channel] += (((changed - images).abs() * masks).sum((1, 2, 3)) - baseline_error).sum().item()
            samples += images.size(0)
    return pd.DataFrame({
        "canal_1_a_N": np.arange(1, channels + 1),
        "indice_python": np.arange(channels),
        "kl_bruta": kl_by_channel.numpy(),
        "participacao_kl_percentual": 100 * kl_by_channel.numpy() / total_kl if total_kl > 0 else np.zeros(channels),
        "variancia_mu_entre_imagens": variances.numpy(),
        "mudanca_media_pixel_ao_zerar": impacts / samples,
        "aumento_erro_l1_pokemon_ao_zerar": foreground_deltas / samples,
    })

channel_analysis_df = analyze_latent_channels(model_spatial, val_loader)
display(channel_analysis_df)
channel_analysis_csv = os.path.join(BASE_DIR, SPATIAL_RUN_NAME + "_canais.csv")
channel_analysis_png = os.path.join(BASE_DIR, SPATIAL_RUN_NAME + "_canais.png")
channel_analysis_df.to_csv(channel_analysis_csv, index=False)
fig_channels, ax_channels = plt.subplots(figsize=(9, 4))
ax_channels.bar(channel_analysis_df["canal_1_a_N"], channel_analysis_df["participacao_kl_percentual"])
ax_channels.set_xticks(channel_analysis_df["canal_1_a_N"])
ax_channels.set_xlabel("Canal latente (1 a N)")
ax_channels.set_ylabel("Participacao na KL total (%)")
ax_channels.set_title(f"{RECONSTRUCTION_KIND.upper()} + Sobel | {model_spatial.latent_channels} canais")
ax_channels.grid(axis="y", alpha=0.3)
fig_channels.tight_layout()
fig_channels.savefig(channel_analysis_png, dpi=150)
plt.show()
CHANNEL_ANALYSIS_PATHS = [channel_analysis_csv, channel_analysis_png]
print("Analise salva e incluida no ZIP final.")
print("KL alta nao prova uso pelo decoder; o teste de zerar complementa a analise.")
print("Zerar pode sair da distribuicao posterior. Os canais nao sao semanticamente alinhados entre modelos.")
```

```python
FIXED_TYPES = ["grass", "poison"]

# Verificar os tipos
assert all(
    pokemon_type in POKEMON_TYPES
    for pokemon_type in FIXED_TYPES
)

# Vetor base com 24 dimensões
base_condition = np.zeros(
    CONDITION_DIM,
    dtype=np.float32
)

# Ativar os tipos
for pokemon_type in FIXED_TYPES:

    type_index = POKEMON_TYPES.index(
        pokemon_type
    )

    base_condition[type_index] = 1.0

# Valores medianos dos status
median_stats = stat_quantiles.loc[0.50, STAT_COLUMNS].values.astype(
    np.float32
)

base_condition[NUM_TYPES:] = median_stats

print("Tipos fixos:", FIXED_TYPES)
print("\nStatus medianos:")

for col, value in zip(STAT_COLUMNS, median_stats):
    print(f"{STAT_LABELS[col]}: {value:.4f}")

print("\nDimensão da condição:", base_condition.shape)
```

```python
variants = []

# Condição base
variants.append({
    "label": "Baseline\nMedian stats",
    "condition": base_condition.copy(),
    "stat": "Baseline",
    "level": "Median"
})

for i, col in enumerate(STAT_COLUMNS):

    # Condição com status baixo
    low_condition = base_condition.copy()

    low_condition[NUM_TYPES + i] = (
        stat_quantiles.loc[0.10, col]
    )

    variants.append({
        "label": f"{STAT_LABELS[col]}\nLow (Q10)",
        "condition": low_condition,
        "stat": STAT_LABELS[col],
        "level": "Low"
    })

    # Condição com status alto
    high_condition = base_condition.copy()

    high_condition[NUM_TYPES + i] = (
        stat_quantiles.loc[0.90, col]
    )

    variants.append({
        "label": f"{STAT_LABELS[col]}\nHigh (Q90)",
        "condition": high_condition,
        "stat": STAT_LABELS[col],
        "level": "High"
    })


conditions_np = np.stack([
    item["condition"]
    for item in variants
])

conditions_test = torch.tensor(
    conditions_np,
    dtype=torch.float32,
    device=DEVICE
)

print("Quantidade de condições:", len(variants))
print("Shape:", conditions_test.shape)

assert conditions_test.shape == (13, 24)
```

```python
# Garantir reprodutibilidade
GENERATION_SEED = 2026

torch.manual_seed(GENERATION_SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(GENERATION_SEED)

# Um único vetor latente espacial
z_fixed = torch.randn(
    1,
    model_cond.latent_channels,
    checkpoint_cond["latent_size"],
    checkpoint_cond["latent_size"],
    device="cpu" if DEVICE.type == "privateuseone" else DEVICE
).to(DEVICE)

# Repetir o mesmo latente para todas as condições
z_repeated = z_fixed.expand(
    len(variants),
    -1,
    -1,
    -1
).contiguous()

model_cond.eval()

with torch.no_grad():

    generated = model_cond.generate(
        conditions_test,
        z=z_repeated
    )

print("Latente utilizado:", z_repeated.shape)
print("Condições:", conditions_test.shape)
print("Imagens geradas:", generated.shape)
```

```python
N_LATENT_SEEDS = 16

multi_seed_results = {
    col: []
    for col in STAT_COLUMNS
}

model_cond.eval()

for seed in range(N_LATENT_SEEDS):

    torch.manual_seed(GENERATION_SEED + seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(
            GENERATION_SEED + seed
        )

    z = torch.randn(
        1,
        model_cond.latent_channels,
        checkpoint_cond["latent_size"],
        checkpoint_cond["latent_size"],
        device="cpu" if DEVICE.type == "privateuseone" else DEVICE
    ).to(DEVICE)

    z_batch = z.expand(
        len(variants),
        -1,
        -1,
        -1
    ).contiguous()

    with torch.no_grad():

        outputs = model_cond.generate(
            conditions_test,
            z=z_batch
        )

    outputs = outputs.squeeze(1)

    for i, col in enumerate(STAT_COLUMNS):

        low_index = 1 + 2 * i
        high_index = 2 + 2 * i

        difference = torch.abs(
            outputs[high_index] - outputs[low_index]
        )

        sensitivity = difference.mean().item()

        multi_seed_results[col].append(
            sensitivity
        )


multi_seed_summary = []

for col in STAT_COLUMNS:

    values = multi_seed_results[col]

    multi_seed_summary.append({

        "Stat": STAT_LABELS[col],

        "Mean Sensitivity": np.mean(values),

        "Std Sensitivity": np.std(values),

        "Min Sensitivity": np.min(values),

        "Max Sensitivity": np.max(values)
    })


multi_seed_df = pd.DataFrame(
    multi_seed_summary
)

display(
    multi_seed_df.style.format({
        "Mean Sensitivity": "{:.5f}",
        "Std Sensitivity": "{:.5f}",
        "Min Sensitivity": "{:.5f}",
        "Max Sensitivity": "{:.5f}"
    })
)
```

## Comparacao do Ultraball original com o experimento selecionado
Esta celula usa validacao e gera diretamente do prior. Nao usa imagens de teste para escolher
hiperparametros. Requer os dois checkpoints: execute a secao Ultraball com `original` e depois
com `ssim_bordas` (ou outra configuracao ajustada), ou mantenha o checkpoint original que ja tiver treinado. A mesma matriz
de ruido e usada nos dois modelos, pois as arquiteturas sao identicas. O objetivo comum da
tabela e calculado com beta=1 para comparar os mesmos termos; ele nao substitui a avaliacao
visual da geracao. Modelos que ignorem z podem ter pouca diversidade mesmo com KL menor.


```python
ORIGINAL_ULTRABALL_PATH = os.path.join(BASE_DIR, "cvae_spatial_16x16x16_best.pt")
# Comparar o original com a configuracao selecionada, sem fixar o experimento 2/80.
ADJUSTED_ULTRABALL_PATH = CHECKPOINT_SPATIAL_PATH

if os.path.abspath(ORIGINAL_ULTRABALL_PATH) == os.path.abspath(ADJUSTED_ULTRABALL_PATH):
    print("Selecione uma configuracao ajustada para comparar com o checkpoint original.")
elif not all(os.path.isfile(p) for p in [ORIGINAL_ULTRABALL_PATH, ADJUSTED_ULTRABALL_PATH]):
    print("Para comparar, disponibilize o checkpoint original e o da configuracao selecionada. Nenhum checkpoint sera substituido aqui.")
else:
    # Quatro combinacoes mais frequentes no treino; atributos medianos somente do treino.
    condition_columns_compare = TYPE_FEATURE_COLUMNS + NORMALIZED_STAT_COLUMNS
    type_rows = train_df[TYPE_FEATURE_COLUMNS].drop_duplicates().copy()
    type_counts = train_df.groupby(TYPE_FEATURE_COLUMNS).size()
    type_rows["count"] = [int(type_counts.loc[tuple(row)]) for row in type_rows.to_numpy()]
    type_rows = type_rows.sort_values("count", ascending=False).head(4)
    compare_vectors = np.concatenate([
        type_rows[TYPE_FEATURE_COLUMNS].to_numpy(dtype=np.float32),
        np.repeat(train_df[NORMALIZED_STAT_COLUMNS].median().to_numpy(dtype=np.float32)[None], len(type_rows), axis=0)
    ], axis=1)
    compare_c = torch.tensor(compare_vectors, device=DEVICE)
    rng_device_compare = "cpu" if DEVICE.type == "privateuseone" else DEVICE
    samples_per_condition = 4
    # Gerador reiniciado por modelo: mesma semente, formas latentes diferentes.
    compare_c = compare_c.repeat_interleave(samples_per_condition, dim=0)
    comparison_kl_rows = []
    for label, path in [("original", ORIGINAL_ULTRABALL_PATH), (SPATIAL_EXPERIMENT, ADJUSTED_ULTRABALL_PATH)]:
        ck_compare = torch.load(path, map_location="cpu", weights_only=True)
        model_compare = CVAESpatial16(latent_channels=ck_compare["latent_channels"],
                                      condition_dim=ck_compare["condition_dim"]).to(DEVICE)
        model_compare.load_state_dict(ck_compare["model_state_dict"])
        model_compare.eval()
        common_metrics = evaluate_spatial(model_compare, val_loader, lambda_bg=0.1, beta=1.0)
        comparison_kl_rows.append({"Configuracao": label, "Beta final": ck_compare["beta_max"],
                                   "Warm-up": ck_compare["beta_warmup_epochs"],
                                   "Lambda bordas": ck_compare.get("lambda_edge", 0.0),
                                   "Val bordas": common_metrics["edge"],
                                   "Val total (beta=1)": common_metrics["total"],
                                   "Val foreground": common_metrics["foreground"],
                                   "Val background": common_metrics["background"],
                                   "Val KL bruta": common_metrics["kl_raw"]})
        generator_compare = torch.Generator(device=rng_device_compare).manual_seed(2026)
        compare_z = torch.randn(samples_per_condition, ck_compare["latent_channels"],
                                ck_compare["latent_size"], ck_compare["latent_size"],
                                generator=generator_compare, device=rng_device_compare).to(DEVICE)
        compare_z = compare_z.repeat(len(type_rows), 1, 1, 1)
        with torch.no_grad():
            samples_compare = model_compare.generate(compare_c, z=compare_z).cpu()
        fig, axes = plt.subplots(len(type_rows), samples_per_condition,
                                 figsize=(10, 3 * len(type_rows)), squeeze=False)
        for i, (_, row) in enumerate(type_rows.iterrows()):
            type_label = " + ".join(t for t in POKEMON_TYPES if row["type_" + t] == 1)
            for j in range(samples_per_condition):
                axes[i, j].imshow(samples_compare[i * samples_per_condition + j, 0],
                                  cmap="gray", vmin=0, vmax=1, interpolation="nearest")
                axes[i, j].axis("off")
                if j == 0:
                    axes[i, j].set_title(type_label)
        fig.suptitle(label + " | geracao pelo prior, sementes fixas")
        plt.tight_layout()
        plt.show()
        del model_compare
    comparison_kl_df = pd.DataFrame(comparison_kl_rows)
    display(comparison_kl_df)
```

## Baixar os resultados desta execucao

Ao executar esta ultima celula no Colab, o download de um ZIP sera iniciado automaticamente.
Ele inclui o melhor checkpoint `.pt`, o historico `.csv` e um resumo da configuracao.
Os nomes sao obtidos do experimento executado: funciona para original, beta 2 / warm-up 40
e beta 2 / warm-up 80, sem precisar editar a celula. Nenhuma imagem do dataset e incluida.

O ZIP nao inclui o notebook aberto e seus graficos: preserve-os tambem por **Arquivo >
Fazer download > Fazer download do .ipynb**. No VS Code, o ZIP e salvo localmente em
`resultados/downloads`, sem iniciar download pelo navegador.


```python
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED
from datetime import datetime, timezone
import json
import hashlib

def export_training_results(checkpoint_path, history_path, output_directory, run_name, config, additional_paths=None):
    sources = [Path(checkpoint_path), Path(history_path)]
    sources.extend(Path(path) for path in (additional_paths or []))
    missing = [str(path) for path in sources if not path.is_file()]
    if missing:
        raise FileNotFoundError("Resultados ausentes; execute o treinamento antes de exportar: " + ", ".join(missing))
    output_directory = Path(output_directory)
    output_directory.mkdir(parents=True, exist_ok=True)
    stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S_%fZ")
    archive_path = output_directory / f"{run_name}_resultados_{stamp}.zip"
    manifest = dict(config)
    manifest["exported_at_utc"] = stamp
    manifest["sha256"] = {}
    for path in sources:
        digest = hashlib.sha256()
        with path.open("rb") as stream:
            for chunk in iter(lambda: stream.read(1024 * 1024), b""):
                digest.update(chunk)
        manifest["sha256"][path.name] = digest.hexdigest()
    with ZipFile(archive_path, "w", compression=ZIP_DEFLATED) as archive:
        for path in sources:
            archive.write(path, arcname=path.name)
        archive.writestr("configuracao.json", json.dumps(manifest, ensure_ascii=False, indent=2))
    return archive_path

export_config = {
    "experiment": SPATIAL_EXPERIMENT,
    "beta_max": float(BETA_MAX),
    "beta_warmup_epochs": int(BETA_WARMUP_EPOCHS),
    "seed": int(SEED),
    "lambda_bg": float(LAMBDA_BG),
    "lambda_edge": float(LAMBDA_EDGE),
    "edge_loss": "sobel_l1_dilated_alpha_mask",
    "selection_metric": "val_l1_masked_plus_kl",
    "reconstruction_kind": RECONSTRUCTION_KIND,
    "reconstruction_details": RECONSTRUCTION_DETAILS,
    "kl_normalization": float(KL_NORMALIZATION),
    "kl_normalization_policy": "fixed_reference_16_channels",
    "latent_channels": int(LATENT_CHANNELS),
    "latent_size": int(LATENT_SIZE),
    "batch_size": int(BATCH_SIZE),
    "epochs_limit": int(EPOCHS),
    "device_name": DEVICE_NAME,
    "torch_version": str(torch.__version__),
}
export_directory = Path("/content") if IN_COLAB else Path(PROJECT_DIR) / "resultados" / "downloads"
results_zip = export_training_results(
    CHECKPOINT_SPATIAL_PATH, HISTORY_SPATIAL_PATH,
    export_directory, SPATIAL_RUN_NAME, export_config,
    additional_paths=CHANNEL_ANALYSIS_PATHS
)
print("ZIP criado:", results_zip)
print("Contem: checkpoint, historico, configuracao e analise dos canais. Baixe o .ipynb separadamente para preservar os graficos.")
if IN_COLAB:
    from google.colab import files
    files.download(str(results_zip))
else:
    print("Resultados disponiveis na pasta local indicada acima.")
```
